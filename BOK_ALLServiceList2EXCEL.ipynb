```python
KEYFROM_USERDATA = True  # 외부에서 입력하려면 False 로
```

필요모듈 임포트. 액셀에 저장할 거라서 액셀도 임포트

```python
import requests
from openpyxl import Workbook
```

키를 입력받는다.

```python
if KEYFROM_USERDATA:
  from google.colab import userdata
  API_KEY = userdata.get('BOK_ECOS')
else:
  import getpass
  api_key = getpass.getpass('Enter API key: ')
```

Base 명령줄을 구성한다.

```python
BASE_URL = "https://ecos.bok.or.kr/api/StatisticTableList"
```

전체 건수를 확인하는 URL구성. 옵션 항목은 생략하면 전체 목록

```python
url = f"{BASE_URL}/{API_KEY}/json/kr/1/1"
```

API를 호출한다.

```python
response = requests.get(url)
```

값을 JSON으로 변환한다.

```python
data = response.json()
```

출력해 본다.

```python
display( data )
```

전체 개수만 따로 추출해 낸다.

```python
total_count = data["StatisticTableList"]["list_total_count"]
```

전체 개수가 잘 추출되었는지 출력해 보자.

```python
print(total_count)
```

이제 전체 목록을 받아 본다. URL을 새로구성

```python
url = f"{BASE_URL}/{API_KEY}/json/kr/1/{total_count}"
```

Open API를 호출한다.

```python
response = requests.get(url)
```

JSON으로 변환한다.

```python
data = response.json()
```

출력해 본다.

```python
display(data)
```

액셀에 저장하기 위해 row 부분만 추출한다.

```python
rows = data["StatisticTableList"]["row"]
```

액셀 파일을 만든다.

```python
wb = Workbook()
ws = wb.active
```

Sheet 이름을 할당한다.

```python
ws.title = "BOK서비스목록"
```

최초 열 3개에 제목을 부여한다.

```python
ws.append(["순번", "코드명", "서비스명"])
```

루프돌면서 순번, 코드명, 서비스명을 입력한다.

```python
for index, row in enumerate(rows, start=1):
    code = row["STAT_CODE"]
    name = row["STAT_NAME"]

    ws.append([index, code, name])
```

이제 액셀 파일로 저장한다. 먼저 파일이름을 지정한다.

```python
file_name = "bok_service_list.xlsx"
```

파일을 저장한다.

```python
wb.save(file_name)
```
