필요한 모듈을 임포트 한다.

```python
import requests
```

명령어 전체를 한 줄로 구성한다.

```python
url = "https://api.open-meteo.com/v1/forecast?latitude=37.5665&longitude=126.9780&current=temperature_2m,relative_humidity_2m,weather_code"
```

API를 호출하여 값을 받아온다.

```python
response = requests.get(url)
```

받아온 값을 찍어본다.

```python
print(response)
```

dir()은 객체의 속성과 메서드 등의 이름 목록을 보여주는 함수이다.

```python
display(dir(response))
```

url에는 어떤 값이 들어있는지 출력해 보자.

```python
print( response.url )
```

text에는 어떤 값이 들어있는지 출력해 보자.

```python
print( response.text )
```

json()은 text의 문자열을 파이썬 객체로 만들어 준다.

```python
data = response.json()
```

display()는 객체를 보기좋게 출력해 준다.

```python
display(data)
```

객체 중 current 영역을 출력해 보자.

```python
display(data["current"])
```

이번엔 명령 부분과 매개변수를 분리해서 호출해 본다.

먼저 명령 부분만 구성한다.

```python
url = "https://api.open-meteo.com/v1/forecast"
```

매개변수 부분만 따로 구성한다.

```python
params = {
    "latitude": 37.5665,
    "longitude": 126.9780,
    "current": "temperature_2m,relative_humidity_2m,weather_code"
}
```

API를 호출한다. 아까 response = requests.get(url)와 비교해보라.

```python
response = requests.get(url, params=params)
```

response의 url부분을 출력해 보자.

```python
print(response.url)
```

데이터를 json 객체로 변환한다.

```python
data = response.json()
```

데이터를 출력해 보자.

```python
display (data)
```
